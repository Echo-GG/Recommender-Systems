```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
p_base = "../Dataset/ml-100k/u1.base"
p_test = "../Dataset/ml-100k/u1.test"

f_base = pd.read_csv(p_base,sep='\t',names=['user_id','item_id','rating','timestamp'])
f_test = pd.read_csv(p_test,sep='\t',names=['user_id','item_id','rating','timestamp'])

print(f"u1.base overview: \n{f_base.head()}\n")
print(f"u1.test overview: \n{f_test.head()}\n")
```

```python
user_ratings = {} # {user_id:[rating1,rating2,...]}
item_ratings = {} # {item_id:[rating1,rating2,...]}

for (u,i,r,ts) in f_base.values:
    if u not in user_ratings:
        user_ratings[u] = []
    user_ratings[u].append(r)
    if i not in item_ratings:
        item_ratings[i] = []
    item_ratings[i].append(r)

print(user_ratings)
print(item_ratings)
```

```python
total = 0
count = 0
for (u,i,r,ts) in f_base.values:
    total += r
    count += 1
mu = total / count
print(mu)
```

```python
user_mean = {} # {user_id:[mui]} mui: mu for item i
# user_ratings = {user_id:[rating1,rating2,...]}
for u_id,ratings in user_ratings.items():
    user_mean[u_id] = sum(ratings) / len(ratings)
print(user_mean)

item_mean = {} # {item_id:[mui]} mui: mu from user u
# item_ratings = {item_id:[rating1,rating2,...]}
for i_id,ratings in item_ratings.items():
    item_mean[i_id] = sum(ratings) / len(ratings)
print(item_mean)
```

```python
b_u = {} # {user_id:[numerator,denominator]}
for u,i,r,ts in f_base.values:
    if u not in b_u:
        b_u[u] = {'numerator':0.0,'count':0}
    b_u[u]['numerator'] += r - item_mean[i]
    b_u[u]['count'] += 1
for u in b_u:
    b_u[u] = b_u[u]['numerator'] / b_u[u]['count']

b_i = {} # {item_id:[numerator,denominator]}
for u,i,r,ts in f_base.values:
    if i not in b_i:
        b_i[i] = {'numerator':0.0,'count':0}
    b_i[i]['numerator'] += r - user_mean[u]
    b_i[i]['count'] += 1
for i in b_i:
    b_i[i] = b_i[i]['numerator'] / b_i[i]['count']
print(b_u)
print(b_i)
```

```python
records = [] # [u,i,actual,prediction1,prediction2,prediction3,prediction4,prediction5,prediction6]

for u,i,r,ts in f_test.values:

    um = user_mean.get(u,mu)
    im = item_mean.get(i,mu)
    bu = b_u.get(u,0.0)
    bi = b_i.get(i,0.0)

    p1 = um
    p2 = im
    p3 = um/2 + im/2
    p4 = bu + im
    p5 = bi + um
    p6 = mu + bu + bi

    records.append((u,i,r,p1,p2,p3,p4,p5,p6))
    
cols = ["user_id","item_id","actual","user_avg","item_avg","user&item avg","user_bias & item_avg","item_bias & user_avg","mu & user_bias & item_bias"]
result = pd.DataFrame(records,columns=cols)
print(result)
```

```python
mae_results = {}
rmse_results = {}
cols = ["user_avg","item_avg","user&item avg","user_bias & item_avg","item_bias & user_avg","mu & user_bias & item_bias"]
for col in cols:
    mae_results[col] =(abs(result[col] - result['actual'])).mean()
    rmse_results[col] = ((result[col] - result['actual'])**2).mean()**0.5

print("MAE:")
for k,v in mae_results.items():
    print(f"{k}: {v:.4f}")

print("\n")

print("RMSE:")
for k,v in rmse_results.items():
    print(f"{k}: {v:.4f}")
```
